You can also set up secrets via Databricks CLI.
https://docs.databricks.com/gcp/en/security/secrets/

The following is for setting via GUI.
- First you need to go to https://<databricks-instance>#secrets/createScope
- Then set up a 1) scope name and choose 2) either "All workspace users" or "Creator"
- In the string_value below, fill out based on what is in your GCP's key.json file. Ensure that the service account has the permission to read data



```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()

# Set your GCP secrets into the scope
w.secrets.put_secret(scope="gcp_scope", key="project_id", string_value="fill_in")
w.secrets.put_secret(scope="gcp_scope", key="client_email", string_value="fill_in")
w.secrets.put_secret(scope="gcp_scope", key="private_key_id", string_value="fill_in")
w.secrets.put_secret(scope="gcp_scope", key="private_key", string_value="fill_in")
```
