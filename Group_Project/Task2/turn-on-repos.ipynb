```python
from databricks.sdk.core import ApiClient
from databricks.sdk.core import Config

cfg = Config(profile='DEFAULT')
client = ApiClient(cfg)
```

First we check what the current value is. `false` means the feature is disabled. "None" or `true` means it is enabled.

```python
client.do("GET", "/api/2.0/workspace-conf?keys=enableProjectTypeInWorkspace")
```

Then we explicitly turn it on.

```python
client.do("PATCH", "/api/2.0/workspace-conf", body={"enableProjectTypeInWorkspace": "true"}, headers={"Content-Type": "application/json"})
```

And now we check that the config value is set to `true`. To disable a repo, check that `enableProjectTypeInWorkspace` is true and set it to `false`.

```python
client.do("GET", "/api/2.0/workspace-conf?keys=enableProjectTypeInWorkspace")
```
