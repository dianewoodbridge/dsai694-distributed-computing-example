```python
import pyspark
```

```python
# Create an RDD using input_file.csv where it returns a distinct faculty RDD in ascending order.
def return_unique_faculty_rdd(file_name:str) -> pyspark.rdd:
    sc = pyspark.SparkContext.getOrCreate()
    rdd = sc.textFile(file_name).map(lambda x : x.split(","))
    faculty_rdd = rdd.map(lambda x: x[0]).distinct().sortBy(lambda x : x)
    return faculty_rdd
```

```python
file_name = "input_file.csv"
faculty_rdd = return_unique_faculty_rdd(file_name)
faculty_rdd.collect()
```

```python
def faculty_course_str(rdd: pyspark.rdd, faculty_name:str) -> str:
    faculty_course = rdd.filter(lambda x: x[0] == faculty_name)\
                        .map(lambda x: x[1])
    return faculty_course.fold("", lambda x,y: x+y)
```

```python
sc = pyspark.SparkContext.getOrCreate()
rdd = sc.textFile(file_name).map(lambda x : x.split(","))
faculty_course_str(rdd, "Diane")
```

```python
def faculty_course_count_list(rdd: pyspark.rdd, faculty_name:str) -> tuple[int, list[str]]:
    faculty_course_rdd = rdd.filter(lambda x : x[0]==faculty_name)
    return faculty_course_rdd.aggregate((0, []),
                              lambda x, y: (x[0] + 1, x[1] + [y[1]]),
                              lambda x, y: (x[0] + y[0], x[1] + y[1]))
```

```python
faculty_course_count_list(rdd, "Diane")
```
